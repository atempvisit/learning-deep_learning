### Explanation of why `z.grad` is empty

In PyTorch autograd:
1. **Leaf Tensors**: Tensors created directly by the user (like `x = torch.tensor(2.0, requires_grad=True)`) are leaf nodes. PyTorch accumulates gradients for these tensors in their `.grad` attribute.
2. **Non-Leaf Tensors**: Intermediate and output tensors (like `y` and `z`) do not accumulate gradients by default to save memory.

Since you called `z.backward()`, the gradient of `z` with respect to the leaf tensor `x` is computed and stored in `x.grad`. Let's check `x.grad` to see the result of $\frac{dz}{dx}$ at $x = 2$.

```python
# Print the gradient of z with respect to the leaf tensor x
print("Gradient of z with respect to x:", x.grad)
```

```python
x = torch.tensor(2.0)
y = torch.tensor(0.0)

w = torch.tensor(1.0)
b = torch.tensor(0.0)
```

```python
def binary_cross_entropy(prediction, target):
  epsilon = 1e-8
  prediction = torch.clamp(prediction, epsilon, 1-epsilon)
  return -(target*torch.log(prediction) + (1 - target) * torch.log(1 - prediction))
```

```python
z = w*x + b
y_pred = torch.sigmoid(z)

loss = binary_cross_entropy(y_pred, y)
loss
```

```python
dloss_dy_pred = (y_pred - y) / (y_pred * (1 - y_pred))
dy_pred_dz = y_pred * (1 - y_pred)
dz_dw = x
dz_db = 1

dl_dw = dloss_dy_pred * dy_pred_dz * dz_dw
dl_db = dloss_dy_pred * dy_pred_dz * dz_db
```

```python
dl_dw
```

```python
dl_db
```
