```python
class Model(nn.Module):
  def __init__(self, num_features):
    super().__init__()

    self.network = nn.Sequential(
        nn.Linear(num_features, 3), # index 0
        nn.ReLU(), # index 1
        nn.Linear(3, 1), # index 2
        nn.Sigmoid() # index 3
    )

  def forward(self, features):
    out = self.network(features)

    return out
```

### What is ReLU?

**ReLU** stands for **Rectified Linear Unit**. It is one of the most popular activation functions used in Deep Learning.

#### How does it work mathematically?
For any input $x$, the ReLU function outputs the input directly if it is positive; otherwise, it outputs zero. Mathematically, it is defined as:

$$f(x) = \max(0, x)$$

* If $x > 0$: the output is $x$
* If $x \le 0$: the output is $0$

#### Why do we use it?
1. **Introduces Non-linearity**: Neural networks need non-linear functions to learn complex patterns. Without activation functions like ReLU, a multi-layer neural network would behave just like a single-layer linear regression model.
2. **Computationally Efficient**: Unlike `sigmoid` or `tanh`, which require expensive exponential calculations, ReLU only requires a simple thresholding operation ($x > 0$).
3. **Reduces Vanishing Gradient Problem**: For positive inputs, the gradient (derivative) of ReLU is always $1$. This helps gradients flow backward through deep layers during training without shrinking too quickly.

```python
import torch
import matplotlib.pyplot as plt

# Generate input values from -10 to 10
x = torch.linspace(-10, 10, 100)

# Apply ReLU activation function
relu = torch.nn.ReLU()
y = relu(x)

# Plot the ReLU function
plt.figure(figsize=(8, 5))
plt.plot(x.numpy(), y.numpy(), label='ReLU(x) = max(0, x)', color='blue', linewidth=2)
plt.title('ReLU Activation Function')
plt.xlabel('Input (x)')
plt.ylabel('Output')
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.legend()
plt.show()
```

```python
class Model(nn.Module):
  def __init__(self, num_features):
    super().__init__()

    self.network = nn.Sequential(
        nn.Linear(num_features, 3),
        nn.ReLU(),
        nn.Linear(3, 1),
        nn.Sigmoid()
    )

  def forward(self, features):
    out = self.network(features)

    return out
```
